```python
def apply_substitution(term, subst):
    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    return (term[0],) + tuple(apply_substitution(x, subst) for x in term[1:])


def occurs(var, term, subst):
    term = apply_substitution(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, x, subst) for x in term[1:])

    return False


def unify(x, y, subst):

    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    # Step 1: If both are identical
    if x == y:
        return subst

    # Step 2: If x is a variable
    if isinstance(x, str) and x.islower():
        if occurs(x, y, subst):
            return None
        subst[x] = y
        return subst

    # Step 3: If y is a variable
    if isinstance(y, str) and y.islower():
        if occurs(y, x, subst):
            return None
        subst[y] = x
        return subst

    # Step 4: Check predicate/function names
    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0]:
            return None

        # Step 5: Check number of arguments
        if len(x) != len(y):
            return None

        # Step 6: Unify corresponding arguments
        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None

exp1 = ("P", "B", "A")
exp2 = ("P", "x", "y")

print("Expression 1:", exp1)
print("Expression 2:", exp2)

result = unify(exp1, exp2, {})

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution Set:", result)
```
