```python

import random

# Calculate number of attacking queen pairs
def heuristic(board):
    count = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                count += 1

    return count


# Hill Climbing Algorithm
def hill_climbing(n):

    # Randomly place queens
    board = [random.randint(0, n - 1) for i in range(n)]

    while True:

        h = heuristic(board)
        print("Board:", board, "Heuristic:", h)

        # Goal state
        if h == 0:
            print("Solution Found!")
            return board

        best_board = board
        best_h = h

        # Generate neighboring states
        for col in range(n):
            for row in range(n):

                new_board = board.copy()
                new_board[col] = row

                new_h = heuristic(new_board)

                if new_h < best_h:
                    best_h = new_h
                    best_board = new_board

        # Check local minimum
        if best_h == h:
            print("Local Minimum Reached!")
            return None

        board = best_board


# Main Program
n = 8
solution = hill_climbing(n)

if solution:
    print("\nFinal Solution:", solution)

    for row in range(n):
        for col in range(n):
            if solution[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()
```
