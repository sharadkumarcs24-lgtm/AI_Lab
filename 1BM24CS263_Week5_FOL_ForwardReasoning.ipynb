```python

# Step 1: Initialize Knowledge Base with facts
facts = [
    ("Human", "Socrates"),
    ("Human", "Plato")
]

# Step 2: Define rules
rules = [
    ("Human", "Mortal")
]

# Step 3: Read the query
query = ("Mortal", "Socrates")

print("Knowledge Base:")
print("Facts:", facts)
print("Rules:", rules)

# Step 4: Forward Chaining
new_facts = True

while new_facts:
    new_facts = False

    for premise, conclusion in rules:
        for fact in facts:

            # Check whether the premise matches the fact
            if premise == fact[0]:

                # Derive a new fact
                derived_fact = (conclusion, fact[1])

                # Add only if it is not already present
                if derived_fact not in facts:
                    facts.append(derived_fact)
                    print("Derived:", derived_fact)
                    new_facts = True

# Step 5: Check whether query is proved
print("\nFinal Knowledge Base:", facts)

if query in facts:
    print("\nQuery Proved:", query)
else:
    print("\nQuery Not Proved:", query)
```
