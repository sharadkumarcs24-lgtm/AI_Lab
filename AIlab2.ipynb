```python
from copy import deepcopy

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

dx = [-1, 1, 0, 0]
dy = [0, 0, -1, 1]

def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

def is_goal(state):
    return state == goal

def print_state(state):
    for row in state:
        print(row)
    print()

def serialize(state):
    return str(state)

def dfs(initial):
    stack = [initial]
    visited = set()
    step = 0

    while stack:
        current = stack.pop()
        step += 1

        print(f"\n Step {step} - Current State:")
        print_state(current)

        if is_goal(current):
            print("Goal State Reached!")
            return

        visited.add(serialize(current))

        x, y = find_zero(current)

        generated = []

        for i in range(4):
            nx, ny = x + dx[i], y + dy[i]

            if 0 <= nx < 3 and 0 <= ny < 3:
                new_state = deepcopy(current)
                new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]

                if serialize(new_state) not in visited:
                    generated.append(new_state)
                    stack.append(new_state)

        print("Generated States:")
        for g in generated:
            print_state(g)

        print("Stack size:", len(stack))

# Initial state
initial_state = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 0, 8]
]

dfs(initial_state)
```

```python
from copy import deepcopy

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

dx = [-1, 1, 0, 0]
dy = [0, 0, -1, 1]

moves = ["UP", "DOWN", "LEFT", "RIGHT"]

def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

def is_goal(state):
    return state == goal

def print_state(state):
    for row in state:
        print(row)
    print()

# DLS with path tracking
def dls(state, depth, limit, path):
    if is_goal(state):
        print("\n GOAL REACHED!\n")
        for step in path:
            print_state(step)
        print("Total steps (blank moves):", len(path) - 1)
        return True

    if depth == limit:
        return False

    x, y = find_zero(state)

    for i in range(4):
        nx, ny = x + dx[i], y + dy[i]

        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = deepcopy(state)
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]

            if new_state not in path:   # avoid cycles in path
                if dls(new_state, depth + 1, limit, path + [new_state]):
                    return True

    return False

# IDS controller
def ids(initial):
    print(" INITIAL STATE")
    print_state(initial)

    limit = 0

    while True:
        print(f"\n DEPTH LIMIT = {limit}")

        if dls(initial, 0, limit, [initial]):
            break

        limit += 1

# Simple initial state
initial_state = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

ids(initial_state)
```
